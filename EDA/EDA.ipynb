```python
df_work = df.copy()
```

```python
df_work["InvoiceDate"] = pd.to_datetime(df_work["InvoiceDate"])
```

```python
df_work["InvoiceDate"].dtype
```

```python
df_work["Month"] = df_work["InvoiceDate"].dt.month
```

```python
df_work[["InvoiceDate", "Month"]].head()
```

```python
df_work["Year"] = df_work["InvoiceDate"].dt.year
```

```python
df_work[["InvoiceDate", "Year", "Month"]].head()
```

```python
df_work["LineValue"] = df_work["Quantity"] * df_work["Price"]
```

```python
df_work[["Quantity","Price","LineValue"]].head()
```

```python
df_work.groupby(["Year","Month"])["LineValue"].sum()
```

```python
import matplotlib.pyplot as plt
monthly_sales = df_work.groupby(
    ["Year", "Month"]
)["LineValue"].sum()

monthly_sales.plot(kind="line", marker="o")
plt.title("Monthly Net Transaction Value")
plt.xlabel("Year, Month")
plt.ylabel("Net Transaction Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df_work.groupby("Country")["LineValue"].sum().sort_values(ascending = False).head(10)
```

```python
df_work.groupby("Country")["Invoice"].nunique().sort_values(ascending = False).head(10)
```

```python
invoice_total = df_work.groupby(["Country","Invoice"])["LineValue"].sum()
avg_invoice_value = invoice_total.groupby(level="Country").mean().sort_values(ascending=False).head(10)
avg_invoice_value
```

```python
country_summary = invoice_total.groupby(level="Country").agg(["mean","count"]).sort_values(by="mean" , ascending=False).head(10)
country_summary
```

```python
filtered_countries = country_summary[country_summary["count"] >= 50] 
filtered_countries.sort_values( by="mean",ascending=False).head(10)
```

```python
country_summary = invoice_total.groupby(
    level="Country"
).agg(["mean", "count"])

country_summary[
    country_summary["count"] >= 50
].sort_values(
    by="mean",
    ascending=False
).head(10)
```

```python
positive_sales = df_work[df_work["Quantity"] > 0 ]
positive_sales.groupby("Description")["Quantity"].sum().sort_values(ascending=False).head(10)
```

```python
most_sales = positive_sales[positive_sales["LineValue"] > 0 ]
most_sales.groupby("Description")["LineValue"].sum().sort_values(ascending=False).head(10)
```
