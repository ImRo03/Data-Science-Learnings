```python
sales = (df_clean[
    (df_clean["Quantity"]>0) & 
    (df_clean["Price"] > 0)]).copy()
sales.head(10)
```

```python
sales["LineValue"] = sales["Quantity"] * sales["Price"]
sales["LineValue"].head(10)
```

```python
sales["YearMonth"] = sales["InvoiceDate"].dt.to_period("M")
sales["YearMonth"]
```

```python
monthly_sales = sales.groupby("YearMonth")["LineValue"].sum()

monthly_sales
```

```python
plt.figure(figsize=(10,8))

plt.plot(
    monthly_sales.index.astype(str),
    monthly_sales.values,
    marker = "o"
)
plt.title("Monthly transaction values")
plt.xlabel("Month")
plt.ylabel("Transaction values")
plt.xticks(rotation= 45)
plt.show()
```

```python
top_sales = (sales.groupby("Description")["Quantity"].sum().sort_values(ascending=False).head(10))
top_sales
```

```python
top_country = (sales.groupby("Country")["LineValue"].sum().sort_values(ascending=False).head())
top_country
```

```python
plt.figure(figsize=(10,8))

sns.barplot(
    x = top_country.values,
    y = top_country.index
)
plt.title("Top 5 countires with high sales")
plt.xlabel("Transaction values")
plt.ylabel("Sales")
plt.tight_layout()
plt.show()
```

```python
top_product_value = (
    sales.groupby("Description")["LineValue"].sum().sort_values(ascending= False).head(5))
top_product_value
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=top_product_value.values,
    y=top_product_value.index
)

plt.title("Top 5 Products by Transaction Value")
plt.xlabel("Transaction Value (£)")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

# My EDA notes

- The original dataset had 525461 rows and 8 columns.
- Missing values: Description had 2928 (0.56%) and Customer ID had 107927 (20.54%).
- All rows with a missing Description had Price = 0.
- I found 6865 exact duplicate rows and removed them in `df_clean`. After that, 518596 rows remained.
- I changed `InvoiceDate` to datetime. After converting it, there were 0 missing/invalid dates.
- There were 12302 rows with negative Quantity. 10181 had invoice numbers starting with C. The other 2121 had Price = 0.
- There were 3681 rows with Price = 0, and 1560 of those had a positive Quantity.
- For sales analysis, I used rows where Quantity > 0 and Price > 0, then calculated `LineValue = Quantity * Price`.

## What I found

- November 2010 had the highest monthly transaction value, about £1,464,293.
- December was much lower at about £437,631, but this dataset ends on 9 December, so it is not a full month.
- The product with the highest quantity was WHITE HANGING HEART T-LIGHT HOLDER, with 58691 units.
- The United Kingdom had the highest transaction value, about £8,812,685.
- `Manual` was the highest description by transaction value, about £262,963, but I should not treat it as a normal product without checking what it means.

These amounts are based on the filtered sales data, so they are transaction values and not final net revenue. I kept the original `df` unchanged and did the cleaning in a separate DataFrame.
