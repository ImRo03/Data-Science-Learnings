```python
import pandas as pd
import numpy as np
df = pd.read_csv("movie.csv", sep=";", encoding="cp1252")
```

## Core patterns

Practice these from memory rather than memorizing long answers.


```python
# Filtering + selecting columns
df.loc[(df["score"] > 8) & (df["votes"] > 100000), ["name", "score", "votes"]]

# Missing values
df["rating_clean"] = df["rating"].fillna("Not Rated")

# Sort + top N + columns
df.sort_values("gross", ascending=False).head(10)[["name", "gross", "budget"]]

# GroupBy + aggregation + sorting
df.groupby("genre", as_index=False)["score"].mean().sort_values("score", ascending=False)

# Top N per group
df.sort_values("score", ascending=False).groupby("genre").head(3)
```
