```python
class LibreryBook:
    def __init__(self,title,author):
        self.title = title
        self.author = author
        self.is_borrowed = False

    def Borrow(self):
        if self.is_borrowed == False:
            self.is_borrowed = True

    def Return_book(self):
        if self.is_borrowed == True:
            self.is_borrowed = False

    def status(self):
        if self.is_borrowed == True:
            return "Borrowed"
        else:
            return "Available"

book = LibreryBook("Atuomic Habit","James clear")
print(book.status())

book.Borrow()
print(book.status())

book.Return_book()
print(book.status())
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price):
        self.items.append((name, price))

    def total(self):
        total = 0

        for item in self.items:
            total += item[1]

        return total

cart = ShoppingCart()

cart.add_item("Mouse", 1000)
cart.add_item("Keyboard", 2000)
cart.add_item("Headphones", 3000)

print(cart.total())
```
