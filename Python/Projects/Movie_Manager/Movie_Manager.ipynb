```python
class Movie:
    def __init__(self,title,genre,year,score):
        self.title = title
        self.genre = genre
        self.year = year
        self.score = score
    def __str__(self):        
        return f"{self.title},{self.genre},{self.year},{self.score}"
        
class MovieManager:
    def __init__(self):
        self.movies = []

    def add_movie(self,movie):
        self.movies.append(movie)

    def show_movies(self):
        for movie in self.movies:
            print(movie)
            
    def search_movies(self,title):
        found = False
        for movie in self.movies:            
            if movie.title == title:
                print(movie)
                found = True
                break
        if not found:
            print("Movie not found")
                
    def filter_by_genre(self,genre):
        for movie in self.movies:
            if movie.genre == genre:
                print(movie)
                
    def highest_rated(self):
        if not self.movies:
            print("No movies available.")
            return
            
        highest_movie = self.movies[0]            
        for movie in self.movies:
            if movie.score >= highest_movie.score:
                highest_movie = movie
        print(highest_movie)

    def remove_movie(self, title):
        found = False

        for movie in self.movies:
            if movie.title == title:
                self.movies.remove(movie)
                found = True
                break
        if not found:
            print("Movie not found.")


movie = Movie("Inception","Sci-Fi",2010,8.8)
movie2 = Movie("The Dark Knight", "Action", 2008, 9.0)
movie3 = Movie("Interstellar", "Sci-Fi", 2014, 8.7)
movie4 = Movie("3 Idiots", "Comedy", 2009, 8.4)                  

manager = MovieManager()

manager.add_movie(movie)
manager.add_movie(movie2)
manager.add_movie(movie3)
manager.add_movie(movie4)

manager.search_movies("Inception")
manager.search_movies("Avatar")

manager.filter_by_genre("Action")
manager.filter_by_genre("Horror")

manager.highest_rated()

manager.remove_movie("Inception")
manager.remove_movie("Avatar")

manager.show_movies()
```
