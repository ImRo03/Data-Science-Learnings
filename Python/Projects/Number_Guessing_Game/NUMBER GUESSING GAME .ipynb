```python
import random #in python we use import function to import from outside
```

```python
random.randint(1,100) #to import we use randint which is integer which can generate auto num between 1,100
```

```python
jackpot = random.randint(1,30)

guess = int(input("guess the num"))

while guess != jackpot:
    if guess < jackpot:
        print("guess higher")
    else:
        print("guess lower")

    guess = int(input("guess the num"))
    
print("correct ans")
```
