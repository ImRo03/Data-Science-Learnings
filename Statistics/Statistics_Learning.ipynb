# Block 1 — Mean, Median, Mode

## Mean
Mean is the arithmetic average.

Formula: Mean = sum of values / number of values.

Example: `10, 20, 30, 40, 50` → Mean = `30`.

Mean is sensitive to extreme values.


```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])
np.mean(data)
```

## Median
Median is the middle value after sorting the data.

For an even number of values, take the average of the two middle values.

Median is useful when extreme values or skewness make the mean less representative.


```python
np.median(data)
```

## Mode
Mode is the most frequently occurring value.

It is especially useful for categorical data such as the most common city, product category, or payment method.


```python
from statistics import mode

mode([10, 20, 20, 30, 40])
```
