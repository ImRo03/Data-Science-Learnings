## Grouped Bar Chart

### Q1 - Average Review Rating by Category and Gender


```python
male_rating = (df[df["Gender"]=="Male"].groupby("Category")["Review Rating"].mean())
female_rating = (df[df["Gender"]=="Female"].groupby("Category")["Review Rating"].mean())

x = np.arange(len(male_rating.index))
width = 0.35

fig,ax = plt.subplots(figsize=(10,10))

ax.bar(x-width/2,male_rating.values,width,label="Male")
ax.bar(x+width/2,female_rating.values,width,label="Female")

ax.set_xticks(x)
ax.set_xticklabels(male_rating.index)

ax.set_title("Avg rating by male and female")
ax.set_xlabel("Category")
ax.set_ylabel("Rating")

ax.legend()
plt.show()
```

### Q2 - Average Purchase Amount by Subscription Status and Category


```python
subs_customer =(df[df["Subscription Status"] == "Yes"].groupby("Category")["Purchase Amount (USD)"].mean())
non_subs_customer = (df[df["Subscription Status"] == "No"].groupby("Category")["Purchase Amount (USD)"].mean())

x = np.arange(len(subs_customer.index))
width = 0.35

fig,ax = plt.subplots(figsize=(10,7))

ax.bar(x - width/2,subs_customer.values,width,label = "Subscribed")
ax.bar(x + width/2,non_subs_customer.values,width,label = "Non-Subscribed")

ax.set_xticks(x)
ax.set_xticklabels(subs_customer.index)

ax.set_title("Subs-non_subs customer and avg")
ax.set_xlabel("Category")
ax.set_ylabel("Purchase Amount (USD)")

ax.legend()
plt.show()

```

## Stacked bar chart

```python
male_count = (
    df[df["Gender"] == "Male"]
    .groupby("Category")
    .size()
)

female_count = (
    df[df["Gender"] == "Female"]
    .groupby("Category")
    .size()
)

fig, ax = plt.subplots(figsize=(9, 5))

ax.bar(
    male_count.index,
    male_count.values,
    label="Male"
)

ax.bar(
    female_count.index,
    female_count.values,
    bottom=male_count.values,
    label="Female"
)

ax.set_title("Customer Composition by Category and Gender")
ax.set_xlabel("Category")
ax.set_ylabel("Number of Customers")

ax.legend()

plt.show()
```

### Q1 - Customer Composition by Payment Method and Subscription Status


```python
subs_customer = (df[df["Subscription Status"] == "Yes"].groupby("Payment Method").size())
non_subs_customer = (df[df["Subscription Status"] == "No"].groupby("Payment Method").size())

fig,ax = plt.subplots(figsize = (10,8))

ax.bar(subs_customer.index,
      subs_customer.values,
      label = "Subscribed"
)
ax.bar(non_subs_customer.index,
       non_subs_customer.values,
       bottom = subs_customer.values,
       label = "Non-subscribed"    
)

ax.set_title("Customer Composition by Payment Method and Subscription Status")
ax.set_xlabel("Payment Method")
ax.set_ylabel("Num of customer")

ax.legend()
plt.show()
```

### Q2 - Customer Composition by Category and Gender
