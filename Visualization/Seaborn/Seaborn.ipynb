## ECDF Plot
An ECDF (Empirical Cumulative Distribution Function) plot shows the proportion of observations that are less than or equal to each value.
- Helps answer questions like “What percentage of customers spent $50 or less?”
- X-axis shows numerical values.
- Y-axis shows cumulative proportion, ranging from 0 to 1.
- Unlike a histogram or KDE, it helps you directly estimate the percentage of observations below a value.

Syntax
sns.ecdfplot(data=df, x="Purchase Amount (USD)")plt.show()

## Regression Plot

- Problem: Understand whether two numerical variables tend to increase or decrease together.
- Why use it? To see the overall trend more easily than with individual points alone.
- Trend line: Summarizes the linear relationship; it does not prove causation.
- Shaded area: Usually shows a confidence interval around the estimated line.
-

## Understanding Correlation Visually

A correlation heatmap gives you a number, while a scatterplot helps you see the relationship between two numerical variables.
- Positive relationship: Points generally move upward from left to right.
- Negative relationship: Points generally move downward from left to right.
- Weak relationship: Points are scattered without a clear straight-line pattern.
- Strength: Points closer to a straight-line pattern indicate a stronger linear relationship. Correlation does not prove causation.

## Joint Plot

A joint plot shows the relationship between two numerical variables in the center and their individual distributions along the edges.
- Use it to study the relationship between two numerical variables and understand each variable's distribution.
- The center usually contains a scatterplot; the edges show distributions.
- x and y select the numerical columns to compare.
- kind controls the central plot, such as "scatter", "reg", or "kde".

## DataFrame-Based Visualization

Combine Pandas filtering with Seaborn to visualize specific customer segments instead of always analysing the entire dataset.
- Use Pandas to select the rows relevant to a business question.
- Pass the filtered DataFrame to Seaborn using data=.
- Choose a chart based on the question you're investigating.
- This helps analyse specific customer groups.

```python
subs_customer = df[df["Subscription Status"] == "Yes"]
sns.histplot(
    data = subs_customer,
    x = "Purchase Amount (USD)"
    
)
plt.show()

female_customer = df[df["Gender"]== "Female"]
sns.histplot(
    data = female_customer,
    x = "Age"
)
plt.show()

high_rating_customer = df[df["Review Rating"] >= 4.0]

sns.histplot(
    data = high_rating_customer,
    x = "Purchase Amount (USD)"
)
plt.show()
```

## Grouped Analysis

Grouped analysis uses Pandas to summarize data by category, then Seaborn to visualize the results.
- groupby() separates rows into groups.
- Calculate a summary, such as the mean, for each group.
- Use Seaborn to compare the summarized values.
- This helps answer business questions such as which customer group spends more on average.
Syntax

```python
avg_spending = df.groupby("Gender",as_index = False)["Purchase Amount (USD)"].mean()
sns.barplot(
    data = avg_spending,
    x  = "Gender",
    y = "Purchase Amount (USD)"
)
plt.show()

Avg_rating = df.groupby("Category",as_index = False)["Review Rating"].mean()
sns.barplot(
    data = Avg_rating,
    x = "Category",
    y = "Review Rating"
)
plt.show()

Subs = df.groupby("Subscription Status",as_index = False)["Previous Purchases"].mean()
sns.barplot(
    data = Subs,
    x = "Subscription Status",
    y = "Previous Purchases"
)
plt.show()
```

## Titles and Labels

Titles and labels make charts easier to understand by explaining what the chart and its axes represent.
- plt.title() adds a chart title.
- plt.xlabel() labels the X-axis.
- plt.ylabel() labels the Y-axis.
- Use clear, meaningful names that explain the business question.

```python
avg_purchase_amount = df.groupby("Category",as_index = False)["Purchase Amount (USD)"].mean()
sns.barplot(
    data = avg_purchase_amount,
    x = "Category",
    y = "Purchase Amount (USD)"
)
plt.title("Avg spending of purchase amount by category")
plt.xlabel("Category")
plt.ylabel("Purchase Amount (USD)")

plt.show()

avg_rating = df.groupby("Gender",as_index = False)["Review Rating"].mean()
sns.barplot(
    data = avg_rating,
    x = "Gender",
    y = "Review Rating"
)
plt.title("avg Rating by gender")
plt.xlabel("Gender")
plt.ylabel("Review Rating")

plt.show()

avg_purchase = df.groupby("Subscription Status",as_index = False)["Previous Purchases"].mean()
sns.barplot(
    data = avg_purchase,
    x = "Subscription Status",
    y = "Previous Purchases"
)
plt.title("Avg Purchases by subscription Status")
plt.xlabel("Subscription Status")
plt.ylabel("Previous Purchases")
plt.show()
```

## Grid and Readability

Grid lines make charts easier to read by helping you estimate values and compare categories.
- sns.set_style() controls the overall visual style.
- "whitegrid" adds a light grid on a white background.
- "darkgrid" adds a grid on a grey background.
- Set the style before creating your chart so it applies consistently.

```python
sns.set_style("whitegrid")

sns.barplot(
    data=avg_spending,
    x="Gender",
    y="Purchase Amount (USD)"
)
plt.show()
```

## Figure Size

Figure size controls the width and height of your chart, making it easier to read.
- figsize sets the chart's width and height.
- Values are specified in inches: (width, height).
- Larger charts can make crowded visualizations easier to understand.
- Set the figure size before creating the plot.

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=avg_spending,
    x="Gender",
    y="Purchase Amount (USD)"
)

plt.show()
```
